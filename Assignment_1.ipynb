Strings Write a Python program to accept a string and:
=>Count the number of characters
=>Count vowels and consonants
=>Reverse the string
=>Check whether it is a palindrome.

```python
s=input("Enter a string: ")
print("Total number of characters: ",len(s))
v,c=0,0
for ch in s:
  if(ch=='a' or ch=='e' or ch=='i' or ch=='o' or ch=='u'):
    v+=1
  else:
    c+=1
print("Total number of vowels: ",v)
print("Total number of consonants: ",c)
rs=s[::-1]
print("Reverse of the string: ",rs)
if(s==rs):
  print("The string is a palindrome")
else:
  print("The string is not a palindrome")
```

Create two sets of integers and demonstrate union, intersection, difference, and symmetric difference. Also create a dictionary containing the names and marks of five students and display the student with the highest marks.

```python
d_a={10,20,30,40,50}
d_b={5,15,25,35,45}
print("Union: ",d_a.union(d_b))
print("Intersection: ",d_a.intersection(d_b))
print("Difference: ",d_a.difference(d_b))
print("Symmetric Difference: ",d_a.symmetric_difference(d_b))
mark={"Manish":95,"Shafia":99,"Himanshu":90,"Manit":91,"Vinayak":92}
stu=""
m=-1
for student,marks in mark.items():
  if(marks>m):
    m=marks
    stu=student
print("Student with highest marks: ",stu)
```

Develop a Python program that accepts the grade/grade point and credit of different courses and calculates the student's SGPA/GPA using:
𝐺𝑃𝐴=Σ(Grade Point×Credit)/ΣCredits

```python
n = int(input("Enter the number of courses: "))

tc = 0
tg = 0

for i in range(1, n + 1):
    print(f"\nCourse {i}")
    credit = float(input("Enter the credit for this course: "))
    grade = float(input("Enter the grade point earned: "))

    tc += credit
    tg += (credit * grade)

gpa = tg / tc
print("\n")
print(f"Total Credits: {tc}")
print(f"Calculated SGPA/GPA: {gpa:.2f}")
```
